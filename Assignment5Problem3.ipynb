```python
import numpy as np
from scipy.stats import ttest_1samp, t

# Recovery times (in days)
old = np.array([
    23.5, 22.3, 20.2, 22.5, 21.3, 23.2, 23.7, 22.7, 20.2, 20.9,
    22.3, 23.1, 20.3, 25.1, 24.8, 22.0, 21.3, 23.2, 21.3, 22.0
])

new = np.array([
    19.5, 20.1, 19.1, 19.2, 21.2, 19.6, 19.8, 20.2, 17.7, 18.6,
    20.8, 21.1, 18.5, 16.1, 17.3, 22.4, 21.5, 18.6, 20.5, 20.4
])

# (a) Compute the difference data (Old - New)
diff = old - new
mean_diff = np.mean(diff)

# (b) Perform one-tailed one-sample t-test (H0: mean_diff <= 0)
t_stat, p_val_two_tailed = ttest_1samp(diff, 0)
p_val_one_tailed = p_val_two_tailed / 2

# (c) Get critical value for one-tailed test (alpha = 0.05)
alpha = 0.05
df = len(diff) - 1
t_critical = t.ppf(1 - alpha, df)

# (d) Conclusion logic
if t_stat > t_critical:
    conclusion = "Reject the null hypothesis: The new drug significantly improves recovery time."
else:
    conclusion = "Fail to reject the null hypothesis: No significant evidence of improvement."

# Print all results
print("Mean Difference (Old - New):", mean_diff)
print("T-Statistic:", t_stat)
print("One-tailed P-Value:", format(p_val_one_tailed, ".6f"))
print("Critical t-Value (df = {}):".format(df), t_critical)
print("Conclusion:", conclusion)
```
